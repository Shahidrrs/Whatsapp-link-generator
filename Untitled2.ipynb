```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os

# Ensure the clean path variable is utilized correctly
dataset_path = os.path.join(path, "OS Collected Data")

print(f"Verifying dataset_path: {dataset_path}")
if not os.path.exists(dataset_path):
    print(f"Error: The directory '{dataset_path}' does not exist.")
else:
    print(f"Success: The directory '{dataset_path}' exists.")
    try:
        contents = os.listdir(dataset_path)
        print(f"Contents of '{dataset_path}': First 5 items: {contents[:5]}")
    except Exception as e:
        print(f"Error checking path contents: {e}")

# High-resolution input processing: Crucial for microscopic bone textures
train_datagen = ImageDataGenerator(
    validation_split=0.2,
    rotation_range=20,
    zoom_range=0.15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.1,
    brightness_range=[0.85, 1.15],
    horizontal_flip=True,
    fill_mode='nearest'
)

# Validation generator stays clean (no distortions to corrupt verification records)
val_datagen = ImageDataGenerator(
    validation_split=0.2
)

# Target resolution scaled from 224 to 300 to provide double the structural clarity
train_gen = train_datagen.flow_from_directory(
    dataset_path,
    target_size=(300, 300),
    batch_size=16,           # Lowered batch size to prevent GPU out-of-memory errors at 300x300
    class_mode='categorical',
    subset='training',
    shuffle=True
)

val_gen = val_datagen.flow_from_directory(
    dataset_path,
    target_size=(300, 300),
    batch_size=16,
    class_mode='categorical',
    subset='validation',
    shuffle=False            # Fixed: Crucial for 1-to-1 alignment with the confusion matrix
)
```

```python
# Cell 13: Model training imports
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB3  # Upgraded to B3 for high-res structural mapping
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout, BatchNormalization, Input
from tensorflow.keras.models import Model
import matplotlib.pyplot as plt
```

```python
# Cell 14: Instantiate Upgraded High-Resolution Core Engine
base_model = tf.keras.applications.EfficientNetB3(
    weights='imagenet',
    include_top=False,
    input_shape=(300, 300, 3)  # Upgraded to 300x300 to match data generators
)

base_model.trainable = False  # Freeze feature extractor for initial training
```

```python
import tensorflow as tf
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout, BatchNormalization
from tensorflow.keras.models import Model

# 1. Setup the functional pipeline matching our 300x300 high-res configuration
inputs = tf.keras.Input(shape=(300, 300, 3))

# EfficientNet models have internal rescaling, but preprocessing ensures perfect tensor alignment
preprocessed_inputs = tf.keras.applications.efficientnet.preprocess_input(inputs)

# 2. Connect the pipeline through our upgraded base_model (initialized in Cell 14)
x = base_model(preprocessed_inputs, training=False)

# 3. Enhanced processing block with regularizers for maximum accuracy
x = GlobalAveragePooling2D()(x)
x = BatchNormalization()(x)
x = Dropout(0.40)(x)  # Prevents co-dependency on specific pixel sets

x = Dense(512, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(x)
x = BatchNormalization()(x)
x = Dropout(0.30)(x)

output = Dense(train_gen.num_classes, activation='softmax')(x)

# 4. Construct the complete final Model
model = Model(inputs=inputs, outputs=output)

# 5. Compile with standard classification head warmup learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import numpy as np
from sklearn.utils import class_weight

# =====================================================================
# 1. CALIBRATED CLASS WEIGHTS
# =====================================================================
class_labels = train_gen.classes
unique_classes = np.unique(class_labels)

raw_weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=unique_classes,
    y=class_labels
)

# Smoothing balance: Keeps Osteopenia highly competitive without flooding precision metrics
smoothed_weights = (raw_weights + 1.0) / 2.0
class_weight_dict = dict(zip(unique_classes, smoothed_weights))
print(f"Calibrated Smoothed Class Weights: {class_weight_dict}")

# =====================================================================
# 2. BUILD HIGH-RESOLUTION MODEL (EfficientNetB3)
# =====================================================================
print("\nInitializing EfficientNetB3 Base Core Engine...")
inputs = layers.Input(shape=(300, 300, 3))

# Initializing B3 for deep macro-structural diagnostic tracking
base_model = tf.keras.applications.EfficientNetB3(
    include_top=False,
    weights='imagenet',
    input_tensor=inputs
)
base_model.trainable = False  # Freeze backbone for initial warmup phase

# Specialized dense processing block with regularization layers
x = layers.GlobalAveragePooling2D()(base_model.output)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.4)(x)
x = layers.Dense(512, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(3, activation='softmax')(x)

model = models.Model(inputs, outputs)

# =====================================================================
# 3. RUNTIME SYSTEM CALLBACKS
# =====================================================================
lr_scheduler = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=2,
    min_lr=1e-7,
    verbose=1
)

early_stop_stage1 = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True,
    verbose=1
)

early_stop_stage2 = EarlyStopping(
    monitor='val_loss',
    patience=7,
    restore_best_weights=True,
    verbose=1
)

# =====================================================================
# 4. STAGE 1: CLASSIFICATION WARMUP
# =====================================================================
print("\nStarting Stage 1: Top-Layer Classification Warmup...")
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Fixed: Assigned output to 'history' variable to track curves completely
history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=12,
    class_weight=class_weight_dict,
    callbacks=[lr_scheduler, early_stop_stage1]
)

# =====================================================================
# 5. STAGE 2: DEEP MULTI-LAYER FINE-TUNING VIA ADAMW
# =====================================================================
print("\n--- Transitioning to Stage 2: Deep 80-Layer Fine-Tuning ---")

base_model.trainable = True
# Unfreeze the top 80 layers of EfficientNetB3 for rich contextual tuning
for layer in base_model.layers[:-80]:
    layer.trainable = False
print("Successfully isolated and unfrozen the top 80 feature layers of EfficientNetB3.")

# Compile using AdamW for strict regularization to stabilize fine-tuning boundaries
model.compile(
    optimizer=tf.keras.optimizers.AdamW(learning_rate=3e-5, weight_decay=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Fine-tuning deep feature blocks under high-resolution conditions...")
history_ft = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=35,
    class_weight=class_weight_dict,
    callbacks=[lr_scheduler, early_stop_stage2]
)

# Save the final optimized architecture weights
model.save("nutribone_highres_effnetB3_model.h5")
print("\n80-90% Target Range Model Training Complete and Saved Successfully!")
```

```python
# Last Cell: Perfectly Sequenced Continuous Evaluation Matrix
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# =====================================================================
# 1. COMBINE AND PLOT CONTINUOUS METRIC TRENDS ACROSS BOTH STAGES
# =====================================================================
plt.figure(figsize=(14, 5))

# Combine Loss History
total_train_loss = list(history.history['loss']) + list(history_ft.history['loss'])
total_val_loss = list(history.history['val_loss']) + list(history_ft.history['val_loss'])

# Combine Accuracy History
total_train_acc = list(history.history['accuracy']) + list(history_ft.history['accuracy'])
total_val_acc = list(history.history['val_accuracy']) + list(history_ft.history['val_accuracy'])

# Plot Loss Progression
plt.subplot(1, 2, 1)
plt.plot(total_train_loss, label='Train Loss (Stage 1 + 2)', color='royalblue')
plt.plot(total_val_loss, label='Val Loss (Stage 1 + 2)', color='orange', linestyle='--')
plt.axvline(x=len(history.history['loss'])-1, color='gray', linestyle=':', label='Fine-Tuning Split')
plt.legend()
plt.title('Complete Multi-Stage Loss Progression')
plt.xlabel('Total Epochs')

# Plot Accuracy Progression
plt.subplot(1, 2, 2)
plt.plot(total_train_acc, label='Train Acc (Stage 1 + 2)', color='royalblue')
plt.plot(total_val_acc, label='Val Acc (Stage 1 + 2)', color='orange', linestyle='--')
plt.axvline(x=len(history.history['accuracy'])-1, color='gray', linestyle=':', label='Fine-Tuning Split')
plt.legend()
plt.title('Complete Multi-Stage Accuracy Progression')
plt.xlabel('Total Epochs')
plt.show()

# =====================================================================
# 2. BULLETPROOF 1-TO-1 ALIGNED PREDICTION BLOCK
# =====================================================================
print("\nExtracting completed model model from disk path...")
eval_model = tf.keras.models.load_model("nutribone_highres_effnetB3_model.h5")

print("Calculating perfectly aligned 1-to-1 validation metrics...")

# Force strict sequential ordering AND clean individual step sizing
val_gen.shuffle = False
val_gen.batch_size = 1   # Completely fixes remaining batch alignment artifacts
val_gen.reset()

# Generate clean, sequential predictions matching disk order
preds = eval_model.predict(val_gen, steps=len(val_gen), verbose=1)
y_pred = np.argmax(preds, axis=1)
y_true = val_gen.classes
class_names = list(val_gen.class_indices.keys())

# Print out your final authentic classification metrics
print("\n--- Real Aligned Classification Report ---")
print(classification_report(y_true, y_pred, target_names=class_names))

# Plot the accurate Confusion Matrix Heatmap
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names, cmap='Blues')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('True Aligned Confusion Matrix')
plt.show()
```
