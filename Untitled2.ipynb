```python
import numpy as np
from PIL import Image
import os

# FIXED PATH: Added the verified wrapper folder name so it hits the images perfectly
train_dataset_path = os.path.join(path, "Knee Osteoarthritis Classification", "train")

categories = ['Normal', 'Osteopenia', 'Osteoporosis']
data = []
labels = []

print("Loading dataset images from:", train_dataset_path)

for category in categories:
    folder_path = os.path.join(train_dataset_path, category)

    # Defensive check to verify the folder path exists
    if not os.path.exists(folder_path):
        print(f"Warning: Folder not found at {folder_path}. Skipping...")
        continue

    # Slicing [:100] protects your system RAM while giving plenty of samples for plotting
    for file in os.listdir(folder_path)[:100]:
        file_path = os.path.join(folder_path, file)
        try:
            # Scaled to native 224x224 to protect RAM and match the new dataset configuration
            img = Image.open(file_path).convert('RGB').resize((224, 224))
            data.append(np.array(img))
            labels.append(category)
        except Exception as e:
            print(f"Error loading {file_path}: {e}")

data = np.array(data)
labels = np.array(labels)
print("\n--- Load Status ---")
print("Data shape:", data.shape)
print("Labels shape:", labels.shape)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os

# FIXED: Added the nested directory layer found by our diagnostic check
base_dataset_folder = os.path.join(path, "Knee Osteoarthritis Classification")

train_dir = os.path.join(base_dataset_folder, "train")
val_dir = os.path.join(base_dataset_folder, "val")
test_dir = os.path.join(base_dataset_folder, "test")

print(f"Successfully located balanced training path: {train_dir}")
print(f"Successfully located balanced validation path: {val_dir}")

# 1. TRAINING GENERATOR: Strong geometric features to keep fighting overfitting
train_datagen = ImageDataGenerator(
    rotation_range=15,
    zoom_range=0.15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.1,
    brightness_range=[0.85, 1.15],
    horizontal_flip=True,
    fill_mode='nearest'
)

# 2. VALIDATION GENERATOR: Completely static
val_datagen = ImageDataGenerator()

# 3. Flow streams configured to the native 224x224 shape
train_gen = train_datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    shuffle=True
)

val_gen = val_datagen.flow_from_directory(
    val_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    shuffle=False  # Keep False for perfect 1-to-1 confusion matrix mapping
)
```

```python
# Cell 12: Model training imports
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB3  # Upgraded to B3 for high-res structural mapping
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout, BatchNormalization, Input
from tensorflow.keras.models import Model
import matplotlib.pyplot as plt
```

```python
# Cell 13: Instantiate Core Engine matched to our 224x224 generator shapes
base_model = tf.keras.applications.EfficientNetB3(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)  # Matches target_size=(224, 224) from Cell 11
)

base_model.trainable = False  # Freeze feature extractor for initial training
```

```python
import tensorflow as tf
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout, BatchNormalization
from tensorflow.keras.models import Model

# 1. FIXED: Set the functional pipeline input shape to 224x224
inputs = tf.keras.Input(shape=(224, 224, 3))

# 2. Connect directly through our frozen base_model
x = base_model(inputs, training=False)

# 3. Enhanced processing block with regularizers for maximum accuracy
x = GlobalAveragePooling2D()(x)
x = BatchNormalization()(x)
x = Dropout(0.40)(x)  # Prevents co-dependency on specific pixel sets

x = Dense(512, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(1e-4))(x)
x = BatchNormalization()(x)
x = Dropout(0.30)(x)

# Automatically extracts output category dimensions from our generator
output = Dense(train_gen.num_classes, activation='softmax')(x)

# 4. Construct the complete final Model
model = Model(inputs=inputs, outputs=output)

# 5. Compile with standard classification head warmup learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# =====================================================================
# 1. BUILD REGULARIZED ENGINE WITH ALIGNED PREPROCESSING (EFFICIENTNETB3)
# =====================================================================
print("\nInitializing EfficientNetB3 Core Engine...")
inputs = layers.Input(shape=(224, 224, 3))

# Load EfficientNetB3 to match our earlier cell architectural choice
base_model = tf.keras.applications.EfficientNetB3(
    include_top=False,
    weights='imagenet',
    input_shape=(224, 224, 3)
)
base_model.trainable = False  # Locked for rapid top-layer warmup

# Connect the pipeline layers smoothly
x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.5)(x)  # Strong dropout to fight memorization

x = layers.Dense(256, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(1e-3))(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.4)(x)
outputs = layers.Dense(3, activation='softmax')(x)

model = models.Model(inputs, outputs)

# =====================================================================
# 2. RUNTIME SYSTEM CALLBACKS (Aggressive Stop Settings)
# =====================================================================
lr_scheduler = ReduceLROnPlateau(
    monitor='val_loss', factor=0.5, patience=2, min_lr=1e-6, verbose=1
)

early_stop_stage1 = EarlyStopping(
    monitor='val_loss', patience=3, restore_best_weights=True, verbose=1
)

early_stop_stage2 = EarlyStopping(
    monitor='val_loss', patience=4, restore_best_weights=True, verbose=1
)

# =====================================================================
# 3. STAGE 1: RAPID CLASSIFICATION WARMUP
# =====================================================================
print("\nStarting Stage 1: Top-Layer Classification Warmup...")
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=10,
    callbacks=[lr_scheduler, early_stop_stage1]
)

# =====================================================================
# 4. STAGE 2: SURGICAL FINE-TUNING VIA ADAMW
# =====================================================================
print("\n--- Transitioning to Stage 2: Deep Fine-Tuning ---")
base_model.trainable = True

# Unfreeze only the top 30 layers for quick, stable optimization boundaries
for layer in base_model.layers[:-30]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.AdamW(learning_rate=3e-5, weight_decay=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Fine-tuning feature blocks smoothly...")
history_ft = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=20,
    callbacks=[lr_scheduler, early_stop_stage2]
)

# Save the final optimized architecture weights under the standard deployment name
model.save("nutribone_highres_effnetB3_model.h5")
print("\nBalanced Model Training Complete and Saved Successfully!")
```

```python
# Last Cell: Perfectly Sequenced Continuous Evaluation Matrix
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# =====================================================================
# 1. COMBINE AND PLOT CONTINUOUS METRIC TRENDS ACROSS BOTH STAGES
# =====================================================================
plt.figure(figsize=(14, 5))

# Combine Loss History safely
total_train_loss = list(history.history['loss']) + list(history_ft.history['loss'])
total_val_loss = list(history.history['val_loss']) + list(history_ft.history['val_loss'])

# FIXED: Fallback handle to protect against Keras 3 'val_accuracy' vs 'val_acc' dictionary key shifts
train_acc_key = 'accuracy' if 'accuracy' in history.history else 'acc'
val_acc_key = 'val_accuracy' if 'val_accuracy' in history.history else 'val_acc'

total_train_acc = list(history.history[train_acc_key]) + list(history_ft.history[train_acc_key])
total_val_acc = list(history.history[val_acc_key]) + list(history_ft.history[val_acc_key])

# Plot Loss Progression
plt.subplot(1, 2, 1)
plt.plot(total_train_loss, label='Train Loss (Stage 1 + 2)', color='royalblue')
plt.plot(total_val_loss, label='Val Loss (Stage 1 + 2)', color='orange', linestyle='--')
plt.axvline(x=len(history.history['loss'])-1, color='gray', linestyle=':', label='Fine-Tuning Split')
plt.legend()
plt.title('Complete Multi-Stage Loss Progression')
plt.xlabel('Total Epochs')

# Plot Accuracy Progression
plt.subplot(1, 2, 2)
plt.plot(total_train_acc, label='Train Acc (Stage 1 + 2)', color='royalblue')
plt.plot(total_val_acc, label='Val Acc (Stage 1 + 2)', color='orange', linestyle='--')
plt.axvline(x=len(history.history[train_acc_key])-1, color='gray', linestyle=':', label='Fine-Tuning Split')
plt.legend()
plt.title('Complete Multi-Stage Accuracy Progression')
plt.xlabel('Total Epochs')
plt.show()

# =====================================================================
# 2. BULLETPROOF 1-TO-1 ALIGNED PREDICTION BLOCK
# =====================================================================
print("\nExtracting completed model from disk path...")
eval_model = tf.keras.models.load_model("nutribone_highres_effnetB3_model.h5")

print("Calculating perfectly aligned 1-to-1 validation metrics...")

# Force strict sequential ordering AND clean individual step sizing
val_gen.shuffle = False
val_gen.batch_size = 1   # Completely fixes remaining batch alignment artifacts
val_gen.reset()

# Generate clean, sequential predictions matching disk order
preds = eval_model.predict(val_gen, steps=len(val_gen), verbose=1)
y_pred = np.argmax(preds, axis=1)
y_true = val_gen.classes
class_names = list(val_gen.class_indices.keys())

# Print out your final authentic classification metrics
print("\n--- Real Aligned Classification Report ---")
print(classification_report(y_true, y_pred, target_names=class_names))

# Plot the accurate Confusion Matrix Heatmap
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names, cmap='Blues')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('True Aligned Confusion Matrix')
plt.show()
```
