```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os

# Ensure the clean path variable is utilized correctly
dataset_path = os.path.join(path, "OS Collected Data")

# 1. TRAINING GENERATOR: Gets full augmentations to force deep feature extraction
train_datagen = ImageDataGenerator(
    validation_split=0.2,
    rotation_range=20,
    zoom_range=0.15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.1,
    brightness_range=[0.85, 1.15],
    horizontal_flip=True,
    fill_mode='nearest'
)

# 2. VALIDATION GENERATOR: Stays completely clean (no random distortions to break evaluation mapping)
val_datagen = ImageDataGenerator(
    validation_split=0.2
)

# --- Start of diagnostic additions ---
print(f"Verifying dataset_path: {dataset_path}")
if not os.path.exists(dataset_path):
    print(f"Error: The directory '{dataset_path}' does not exist.")
else:
    print(f"Success: The directory '{dataset_path}' exists.")
    try:
        # Attempt to list directory contents to further diagnose access issues
        contents = os.listdir(dataset_path)
        print(f"Contents of '{dataset_path}': First 5 items: {contents[:5]}")
        if not any(os.path.isdir(os.path.join(dataset_path, item)) for item in contents):
            print("Warning: No subdirectories found in dataset_path. `flow_from_directory` expects class subfolders.")
    except FileNotFoundError as e:
        print(f"Error: Still unable to list contents of '{dataset_path}'. Original error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred while trying to list directory contents: {e}")
# --- End of diagnostic additions ---

# Stream training images from the heavily augmented generator configuration
train_gen = train_datagen.flow_from_directory(
    dataset_path,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='training'
)

# Stream validation images from the completely clean, static generator configuration
val_gen = val_datagen.flow_from_directory(
    dataset_path,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)
```

```python
import tensorflow as tf
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model

# 1. Initialize the core pre-trained base model
base = EfficientNetB0(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)
base.trainable = False   # Keep features locked to prevent unstable gradients

# 2. Setup the functional pipeline with automatic image normalization
inputs = tf.keras.Input(shape=(224, 224, 3))
preprocessed_inputs = tf.keras.applications.efficientnet.preprocess_input(inputs)

# 3. Connect the pipeline through the base and your updated classification head
x = base(preprocessed_inputs, training=False)
x = GlobalAveragePooling2D()(x)
x = Dropout(0.5)(x)              # Safe regularization factor
x = Dense(256, activation='relu')(x)
x = Dropout(0.5)(x)              # Safe regularization factor
output = Dense(train_gen.num_classes, activation='softmax')(x)

# 4. Construct the complete final Model
model = Model(inputs=inputs, outputs=output)

# 5. Compile with the stable medical training learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import numpy as np
from sklearn.utils import class_weight

# =====================================================================
# 1. CALCULATE SMOOTHED CLASS WEIGHTS
# =====================================================================
class_labels = train_gen.classes
unique_classes = np.unique(class_labels)

raw_weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=unique_classes,
    y=class_labels
)

# Apply mathematical smoothing to prevent over-biasing the model
smoothed_weights = (raw_weights + 1.0) / 2.0
class_weight_dict = dict(zip(unique_classes, smoothed_weights))
print(f"Smoothed Class Weights (Calibrated): {class_weight_dict}")

# =====================================================================
# 2. DEFINE STABILIZED CALLBACKS
# =====================================================================
lr_scheduler = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=3,
    min_lr=1e-7,
    verbose=1
)

early_stop_stage1 = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True,
    verbose=1
)

early_stop_stage2 = EarlyStopping(
    monitor='val_loss',
    patience=8,
    restore_best_weights=True,
    verbose=1
)

# =====================================================================
# 3. STABLE INITIAL TRAINING (Stage 1)
# =====================================================================
print("\nStarting initial training loop (Stage 1)...")
history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=12,
    class_weight=class_weight_dict,
    callbacks=[lr_scheduler, early_stop_stage1]
)

# =====================================================================
# 4. DEEPER FINE-TUNING VIA ADAMW (Stage 2)
# =====================================================================
print("\n--- Transitioning to Stage 2: Deep 50-Layer Fine-Tuning ---")

efficientnet_base = None
for layer in model.layers:
    if "efficientnet" in layer.name.lower():
        efficientnet_base = layer
        break

if efficientnet_base is not None:
    efficientnet_base.trainable = True
    # Unfreeze the top 50 layers instead of 20 to give the model more structural flexibility
    for layer in efficientnet_base.layers[:-50]:
        layer.trainable = False
    print("Successfully isolated and unfrozen the top 50 layers of EfficientNet.")
else:
    base.trainable = True
    for layer in base.layers[:-50]:
        layer.trainable = False

# Recompile using AdamW to enforce weight regularization during deep fine-tuning
model.compile(
    optimizer=tf.keras.optimizers.AdamW(learning_rate=3e-5, weight_decay=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Fine-tuning expanded feature blocks under AdamW optimization...")
history_ft = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=35,
    class_weight=class_weight_dict,
    callbacks=[lr_scheduler, early_stop_stage2]
)

# Save the final optimized weights
model.save("nutribone_effnet_model.h5")
print("\nMaximum Accuracy Fine-Tuned Model Saved Locally!")
```

```python
# Stage 2: Perfectly Sequenced Evaluation
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# =====================================================================
# 1. PLOT METRIC TRENDS FROM MODEL TRAINING
# =====================================================================
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.legend(); plt.title('Loss Progression')

plt.subplot(1,2,2)
plt.plot(history.history['accuracy'], label='train_acc')
plt.plot(history.history['val_accuracy'], label='val_acc')
plt.legend(); plt.title('Accuracy Progression')
plt.show()

# =====================================================================
# 2. BULLETPROOF 1-TO-1 ALIGNED PREDICTION BLOCK
# =====================================================================
print("\nCalculating perfectly aligned 1-to-1 validation metrics...")

# 1. Force sequential ordering AND clean individual step sizing
val_gen.shuffle = False
val_gen.batch_size = 1   # Fixes the remainder batch mismatch
val_gen.reset()

# 2. Generate clean, sequential predictions matching disk order
preds = model.predict(val_gen, steps=len(val_gen), verbose=1)
y_pred = np.argmax(preds, axis=1)
y_true = val_gen.classes
class_names = list(val_gen.class_indices.keys())

# 3. Print your authentic classification metrics
print("\n--- Real Aligned Classification Report ---")
print(classification_report(y_true, y_pred, target_names=class_names))

# 4. Plot the accurate Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names, cmap='Blues')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('True Aligned Confusion Matrix')
plt.show()
```
