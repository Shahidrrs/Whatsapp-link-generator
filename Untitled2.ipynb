```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os
dataset_path = os.path.join(path, "OS Collected Data")

train_datagen = ImageDataGenerator(
    validation_split=0.2,
    rotation_range=15,
    zoom_range=0.1,
    horizontal_flip=True
)

# --- Start of diagnostic additions ---
print(f"Verifying dataset_path: {dataset_path}")
if not os.path.exists(dataset_path):
    print(f"Error: The directory '{dataset_path}' does not exist.")
else:
    print(f"Success: The directory '{dataset_path}' exists.")
    try:
        # Attempt to list directory contents to further diagnose access issues
        contents = os.listdir(dataset_path)
        print(f"Contents of '{dataset_path}': First 5 items: {contents[:5]}")
        if not any(os.path.isdir(os.path.join(dataset_path, item)) for item in contents):
            print("Warning: No subdirectories found in dataset_path. `flow_from_directory` expects class subfolders.")
    except FileNotFoundError as e:
        print(f"Error: Still unable to list contents of '{dataset_path}'. Original error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred while trying to list directory contents: {e}")
# --- End of diagnostic additions ---

train_gen = train_datagen.flow_from_directory(
    dataset_path,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='training'
)

val_gen = train_datagen.flow_from_directory(
    dataset_path,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)
```

```python
#model traning
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model
import matplotlib.pyplot as plt
```

```python
import tensorflow as tf
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model

# 1. Initialize the core pre-trained base model
base = EfficientNetB0(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)
base.trainable = False   # Keep features locked to prevent unstable gradients

# 2. Setup the functional pipeline with automatic image normalization
inputs = tf.keras.Input(shape=(224, 224, 3))
preprocessed_inputs = tf.keras.applications.efficientnet.preprocess_input(inputs)

# 3. Connect the pipeline through the base and your updated classification head
x = base(preprocessed_inputs, training=False)
x = GlobalAveragePooling2D()(x)
x = Dropout(0.5)(x)              # Safe regularization factor
x = Dense(256, activation='relu')(x)
x = Dropout(0.5)(x)              # Safe regularization factor
output = Dense(train_gen.num_classes, activation='softmax')(x)

# 4. Construct the complete final Model
model = Model(inputs=inputs, outputs=output)

# 5. Compile with the stable medical training learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# =====================================================================
# STEP 1: DEFINE CALLBACKS FOR STAGE 1
# =====================================================================
lr_scheduler = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

early_stop_stage1 = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True,
    verbose=1
)

# =====================================================================
# STEP 2: STABLE INITIAL TRAINING (Stage 1)
# =====================================================================
print("Starting initial training loop (Stage 1)...")
history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=12,
    callbacks=[lr_scheduler, early_stop_stage1]
)

# =====================================================================
# STEP 3: SURGICAL FINE-TUNING (Stage 2)
# =====================================================================
print("\n--- Transitioning to Stage 2: Partial Fine-Tuning ---")

efficientnet_base = None
for layer in model.layers:
    if "efficientnet" in layer.name.lower():
        efficientnet_base = layer
        break

if efficientnet_base is not None:
    efficientnet_base.trainable = True
    for layer in efficientnet_base.layers[:-20]:
        layer.trainable = False
    print("Successfully isolated and unfrozen the top 20 layers of EfficientNet.")
else:
    base.trainable = True
    for layer in base.layers[:-20]:
        layer.trainable = False

# Recompile with a slightly more active fine-tuning learning rate (5e-5)
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=5e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Force the model to run through the fine-tuning paces cleanly
print("Fine-tuning top feature blocks...")
history_ft = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=15,
    callbacks=[lr_scheduler] # Removed hard early stop to allow adaptation
)

# Save the final optimized model weights
model.save("nutribone_effnet_model.h5")
print("\nOptimized Fine-Tuned Model Saved Locally!")
```

```python
# Stage 2: Balanced Model Evaluation
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# =====================================================================
# 1. PLOT METRIC TRENDS FROM MODEL TRAINING
# =====================================================================
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.legend(); plt.title('Loss Progression')

plt.subplot(1,2,2)
plt.plot(history.history['accuracy'], label='train_acc')
plt.plot(history.history['val_accuracy'], label='val_acc')
plt.legend(); plt.title('Accuracy Progression')
plt.show()

# =====================================================================
# 2. ALIGNED VALIDATION EVALUATION (Fixes Generator Shuffling Bug)
# =====================================================================
print("\nCalculating aligned validation metrics...")

# Force the data generator to be completely sequential for testing
val_gen.shuffle = False
val_gen.reset()

# Generate fresh predictions matching the raw sequence on disk
preds = model.predict(val_gen, verbose=1)
y_pred = np.argmax(preds, axis=1)
y_true = val_gen.classes
class_names = list(val_gen.class_indices.keys())

# Print out your true, aligned performance metrics
print("\n--- Aligned Classification Report ---")
print(classification_report(y_true, y_pred, target_names=class_names))

# Plot the real, un-shuffled Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names, cmap='Blues')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Real Aligned Confusion Matrix')
plt.show()
```
